```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("C:\\Users\\Brindha Suvarna R\\Desktop\\credit-risk-modeling\\data\\raw\\application_train.csv")

df.shape
```

```python
df_clean = df.copy()
```

```python
cols_to_drop = [
'COMMONAREA_AVG','COMMONAREA_MODE','COMMONAREA_MEDI',
'NONLIVINGAPARTMENTS_MEDI','NONLIVINGAPARTMENTS_MODE','NONLIVINGAPARTMENTS_AVG',
'FONDKAPREMONT_MODE',
'LIVINGAPARTMENTS_AVG','LIVINGAPARTMENTS_MEDI','LIVINGAPARTMENTS_MODE',
'FLOORSMIN_MODE','FLOORSMIN_AVG','FLOORSMIN_MEDI',
'YEARS_BUILD_AVG','YEARS_BUILD_MODE','YEARS_BUILD_MEDI',
'OWN_CAR_AGE',
'LANDAREA_MEDI','LANDAREA_AVG','LANDAREA_MODE'
]

df_clean.drop(columns=cols_to_drop, inplace=True)

df_clean.shape
```

```python
df_clean['DAYS_EMPLOYED'].replace(365243, np.nan, inplace=True)
```

```python
num_cols = df_clean.select_dtypes(include=['int64','float64']).columns
cat_cols = df_clean.select_dtypes(include=['object']).columns

print("Numerical columns:", len(num_cols))
print("Categorical columns:", len(cat_cols))
```

```python
for col in num_cols:
    df_clean[col].fillna(df_clean[col].median(), inplace=True)
```

```python
for col in cat_cols:
    df_clean[col].fillna(df_clean[col].mode()[0], inplace=True)
```
