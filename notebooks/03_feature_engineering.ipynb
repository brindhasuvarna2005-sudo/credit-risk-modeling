```python
import pandas as pd

df = pd.read_csv("../data/processed/clean_application_data.csv")
```

```python
df.shape
```

```python
df.columns
```

```python
df["AGE_YEARS"] = abs(df["DAYS_BIRTH"]) / 365
df["EMPLOYMENT_YEARS"] = abs(df["DAYS_EMPLOYED"]) / 365
```

```python
df["CREDIT_TO_INCOME_RATIO"] = df["AMT_CREDIT"] / df["AMT_INCOME_TOTAL"]

df["ANNUITY_TO_INCOME_RATIO"] = df["AMT_ANNUITY"] / df["AMT_INCOME_TOTAL"]

df["CREDIT_TO_ANNUITY_RATIO"] = df["AMT_CREDIT"] / df["AMT_ANNUITY"]

df["INCOME_PER_PERSON"] = df["AMT_INCOME_TOTAL"] / df["CNT_FAM_MEMBERS"]
```

```python
df["CHILDREN_RATIO"] = df["CNT_CHILDREN"] / df["CNT_FAM_MEMBERS"]
```

```python
df["EMPLOYED_TO_AGE_RATIO"] = df["EMPLOYMENT_YEARS"] / df["AGE_YEARS"]
```

```python
df["TOTAL_CREDIT_INQUIRIES"] = (
    df["AMT_REQ_CREDIT_BUREAU_HOUR"]
    + df["AMT_REQ_CREDIT_BUREAU_DAY"]
    + df["AMT_REQ_CREDIT_BUREAU_WEEK"]
    + df["AMT_REQ_CREDIT_BUREAU_MON"]
    + df["AMT_REQ_CREDIT_BUREAU_QRT"]
    + df["AMT_REQ_CREDIT_BUREAU_YEAR"]
)
```

```python
df = pd.get_dummies(df, drop_first=True)
```
